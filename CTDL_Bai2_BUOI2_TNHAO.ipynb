```python
txt = """
=====MENU=====
1.Lấy số thứ tự
2.Gọi người tiếp theo-
3.Xem người đầu hàng
4.Số lượng đang chờ
5.Trạng thái hàng đợi"""
print(txt)

from collections import deque

class Hangdoisinhvien:
  def __init__(self):
    self.queue = deque()
    self.dem = 0
    self.lich_su = []

  def ThongTin(self, thao_tac):
    trang_thai = list(self.queue)
    self.lich_su.append((thao_tac, trang_thai.copy()))
    print(f"{thao_tac} - Hàng đợi: {trang_thai} - Đang chờ: {len(self.queue)}")

  def LaySo(self):
    self.dem += 1
    sv = f"SV{self.dem}"
    self.queue.append(sv)
    self.ThongTin(f"Lấy số: {sv}")
    return sv

  def GoiNguoi(self):
    if not self.queue:
      self.ThongTin("Gọi tiếp theo: (hàng đợi rỗng)")
      return None
    sv = self.queue.popleft()
    self.ThongTin(f"Đang phục vụ: {sv}")
    return sv

  def XemNguoiDauHang(self):
    if not self.queue:
      print("Người đầu hàng: (không có)")
      self.ThongTin(f"Xem đầu hàng: rỗng")
      return None
    sv = self.queue[0]
    print("Người đầu hàng: ", sv)
    self.ThongTin(f"Xem người đầu hàng: {sv}")
    return sv

  def SoLuongCho(self):
    n = len(self.queue)
    print("Số lượng đang chờ", n)
    self.ThongTin(f"Xem số lượng đang chờ: {n}")
    return n

  def XemLichSu(self):
    if not self.lich_su:
      print("Chưa có thao tác nào")
    for i, (tt, state) in enumerate(self.lich_su, 1):
      print(f"{i}. {tt} -> {state}")

hd = Hangdoisinhvien()

while True:
  chon = input("Chọn chức năng: ").strip()
  if chon == '1': hd.LaySo()
  elif chon == '2': hd.GoiNguoi()
  elif chon == '3': hd.XemNguoiDauHang()
  elif chon == '4': hd.SoLuongCho()
  elif chon == '5': hd.XemLichSu()
  elif chon == '0':
    print("Kết thúc chương trình")
    break
  else:
    print("Lựa chọn không hợp lệ")
```
