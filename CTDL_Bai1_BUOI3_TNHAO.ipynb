```python
class TreeNode:
  def __init__(self, gia_tri):
    self.gia_tri = gia_tri
    self.trai = None
    self.phai = None
def duyet_truoc(goc):
  if goc is not None:
    print(goc.gia_tri, end=" ")
    duyet_truoc(goc.trai)
    duyet_truoc(goc.phai)
def duyet_giua(goc):
  if goc is not None:
    duyet_giua(goc.trai)
    print(goc.gia_tri, end=" ")
    duyet_giua(goc.phai)
def duyet_sau(goc):
  if goc is not None:
    duyet_sau(goc.trai)
    duyet_sau(goc.phai)
    print(goc.gia_tri, end=" ")

goc = TreeNode(10)
goc.trai = TreeNode(5)
goc.phai = TreeNode(20)
goc.trai.trai = TreeNode(2)
goc.trai.phai = TreeNode(8)

print("Tiền thứ tự: ", end="")
duyet_truoc(goc)
print()

print("Trung thứ tự: ", end="")
duyet_giua(goc)
print()

print("Hậu thứ tự: ", end="")
duyet_sau(goc)
print()
```
