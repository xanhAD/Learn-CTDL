```python
danh_sach_sv = []

def them_sv():
    ma_sv = input("Nhập mã sinh viên: ").strip()
    if not ma_sv:
        print("Mã sinh viên không được để trống")
        return
    for sv in danh_sach_sv:
        if sv['ma_sv'] == ma_sv:
            print("Mã sinh viên không thể trùng lặp")
            return
    ten_sv = input("Nhập tên sinh viên: ")
    while True:
        try:
            diem = float(input("Nhập điểm sinh viên: "))
            if 0.0 <= diem <= 10.0:
                break
            print("Điểm từ (0.0 - 10.0)")
        except ValueError:
            print("Điểm không hợp lệ")
    sinh_vien = {
        "ma_sv": ma_sv,
        "ten_sv": ten_sv,
        "diem": diem
    }
    danh_sach_sv.append(sinh_vien)
    print("Đã thêm sinh viên thành công!")

def hien_thi_danh_sach_sv():
    if not danh_sach_sv:
        print("Chưa có danh sách sinh viên")
        return
    for sv in danh_sach_sv:
        print(f"Mã: {sv['ma_sv']} | Tên: {sv['ten_sv']} | Điểm: {sv['diem']}")

def tim_theo_ma_sv():
    search = input("Nhập mã sv: ").strip()
    found = False
    for sv in danh_sach_sv:
        if sv['ma_sv'] == search:
            print(f"Mã: {sv['ma_sv']} | Tên: {sv['ten_sv']} | Điểm: {sv['diem']}")
            found = True
            break
    if not found:
        print("Không tìm thấy sinh viên có mã này")

def xoa_theo_ma_sv():
    search = input("Nhập mã sv: ").strip()
    for sv in danh_sach_sv:
        if sv['ma_sv'] == search:
            danh_sach_sv.remove(sv)
            print(f"Đã xóa sinh viên có mã: {search}")
            return
    print("Không tìm thấy sinh viên có mã này để xóa")

def thong_ke_sv():
    if not danh_sach_sv:
        print("Chưa có danh sách sinh viên")
        return
    print("Sinh viên có điểm >= 5:")
    for sv in danh_sach_sv:
        if sv['diem'] >= 5:
            print(f"Mã: {sv['ma_sv']} | Tên: {sv['ten_sv']} | Điểm: {sv['diem']}")

    max_diem = max(sv['diem'] for sv in danh_sach_sv)
    print(f"\nSinh viên có điểm cao nhất ({max_diem}):")
    for sv in danh_sach_sv:
        if sv['diem'] == max_diem:
            print(f"Mã: {sv['ma_sv']} | Tên: {sv['ten_sv']} | Điểm: {sv['diem']}")

def sap_xep():
    danh_sach_sv.sort(key=lambda sv: (-sv['diem'], sv['ma_sv']))
    print("Đã sắp xếp danh sách sinh viên theo yêu cầu:")
    hien_thi_danh_sach_sv()

txt = """===Quản lý sinh viên===
1. Thêm sinh viên
2. Hiển thị danh sách sinh viên
3. Tìm theo mã sinh viên
4. Xóa theo mã sinh viên
5. Thống kê sinh viên
6. Sắp xếp danh sách sinh viên
0. Thoát"""
print(txt)
while True:
    print("============================")
    chon = input("Chọn chức năng: ").strip()
    if    chon == "1": them_sv()
    elif  chon == "2": hien_thi_danh_sach_sv()
    elif  chon == "3": tim_theo_ma_sv()
    elif  chon == "4": xoa_theo_ma_sv()
    elif  chon == "5": thong_ke_sv()
    elif  chon == "6": sap_xep()
    elif  chon == '0': break
    else: print("Lựa chọn không hợp lệ")
```
