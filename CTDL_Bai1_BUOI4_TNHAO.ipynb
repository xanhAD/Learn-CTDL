```python
class BSTNode:
    def __init__(self, key, data=None):
        self.key = key
        self.data = data
        self.left = None
        self.right = None

def insert(root, key, data=None):
    if root is None:
        return BSTNode(key, data)
    if key < root.key:
        root.left = insert(root.left, key, data)
    elif key > root.key:
        root.right = insert(root.right, key, data)
    else:
        raise ValueError("Khóa đã tồn tại")
    return root

def inorder(root, result=None):
    if result is None:
        result = []
    if root:
        inorder(root.left, result)
        result.append(root.key)
        inorder(root.right, result)
    return result

def search_with_count(root, key):
    count = 0
    current = root
    while current is not None:
        count += 1
        if key == current.key:
            return current, count
        elif key < current.key:
            current = current.left
        else:
            current = current.right
    return None, count

def find_min(root):
    if root is None:
        return None
    current = root
    while current.left is not None:
        current = current.left
    return current.key

def find_max(root):
    if root is None:
        return None
    current = root
    while current.right is not None:
        current = current.right
    return current.key

keys = [45, 15, 75, 10, 35, 55, 90]
root = None
for k in keys:
    root = insert(root, k)

print("1. Danh sách đã chèn:", keys)

inorder_list = inorder(root)
print("2. Duyệt inorder:", inorder_list)
print("   Xác nhận tăng dần:", inorder_list == sorted(inorder_list))

print("3. Kết quả tìm kiếm:")
for k in [10, 55, 100]:
    node, count = search_with_count(root, k)
    status = "Tìm thấy" if node else "Không tìm thấy"
    print(f"   Khóa {k}: {status}, số node đi qua: {count}")

print(f"4. Khóa nhỏ nhất: {find_min(root)}")
print(f"   Khóa lớn nhất: {find_max(root)}")
```
