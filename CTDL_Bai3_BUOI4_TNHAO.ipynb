```python
class Node:
    def __init__(self, ma_sv, ho_ten, diem):
        self.ma_sv = ma_sv
        self.ho_ten = ho_ten
        self.diem = diem
        self.left = None
        self.right = None

class StudentBST:
    def __init__(self):
        self.root = None

    def insert(self, ma_sv, ho_ten, diem):
        self.root = self._insert(self.root, ma_sv, ho_ten, diem)

    def _insert(self, root, ma_sv, ho_ten, diem):
        if root is None:
            return Node(ma_sv, ho_ten, diem)
        if ma_sv < root.ma_sv:
            root.left = self._insert(root.left, ma_sv, ho_ten, diem)
        elif ma_sv > root.ma_sv:
            root.right = self._insert(root.right, ma_sv, ho_ten, diem)
        return root

    def search(self, ma_sv):
        steps = 0
        curr = self.root
        while curr:
            steps += 1
            if curr.ma_sv == ma_sv:
                return curr, steps
            elif ma_sv < curr.ma_sv:
                curr = curr.left
            else:
                curr = curr.right
        return None, steps

    def _min_value_node(self, node):
        current = node
        while current.left is not None:
            current = current.left
        return current

    def delete(self, ma_sv):
        self.root = self._delete(self.root, ma_sv)

    def _delete(self, root, ma_sv):
        if root is None:
            return root
        if ma_sv < root.ma_sv:
            root.left = self._delete(root.left, ma_sv)
        elif ma_sv > root.ma_sv:
            root.right = self._delete(root.right, ma_sv)
        else:
            if root.left is None:
                return root.right
            elif root.right is None:
                return root.left
            temp = self._min_value_node(root.right)
            root.ma_sv = temp.ma_sv
            root.ho_ten = temp.ho_ten
            root.diem = temp.diem
            root.right = self._delete(root.right, temp.ma_sv)
        return root

    def inorder(self):
        res = []
        self._inorder(self.root, res)
        return res

    def _inorder(self, root, res):
        if root:
            self._inorder(root.left, res)
            res.append(root)
            self._inorder(root.right, res)

    def count_and_sum(self, root):
        if root is None:
            return 0, 0
        c_l, s_l = self.count_and_sum(root.left)
        c_r, s_r = self.count_and_sum(root.right)
        return 1 + c_l + c_r, root.diem + s_l + s_r

    def get_stats(self):
        count, total = self.count_and_sum(self.root)
        avg = total / count if count > 0 else 0
        return count, avg

def linear_search(lst, ma_sv):
    steps = 0
    for item in lst:
        steps += 1
        if item[0] == ma_sv:
            return item, steps
    return None, steps

bst = StudentBST()
data = [
    ("SV03", "Nguyen An", 8.5),
    ("SV01", "Tran Binh", 7.0),
    ("SV04", "Le Chi", 9.0),
    ("SV02", "Pham Duong", 6.5)
]

for sv in data:
    bst.insert(*sv)

print("Danh sach sinh vien tang dan:")
for sv in bst.inorder():
    print(f"{sv.ma_sv} - {sv.ho_ten} - {sv.diem}")

cnt, avg = bst.get_stats()
print(f"Tong so SV: {cnt}, Diem trung binh: {avg:.2f}")

target = "SV02"
node_bst, bst_steps = bst.search(target)
item_lst, lst_steps = linear_search(data, target)

print(f"Tim kiem {target}:")
print(f"BST: {bst_steps} buoc")
print(f"Linear List: {lst_steps} buoc")
```
