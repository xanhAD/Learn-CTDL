```python
graph = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A", "D"],
    "D": ["B", "C", "E"],
    "E": ["D"]
}

vertices = sorted(list(graph.keys()))
n = len(vertices)
vertex_to_index = {v: i for i, v in enumerate(vertices)}

matrix = [[0] * n for _ in range(n)]
for u, neighbors in graph.items():
    for v in neighbors:
        matrix[vertex_to_index[u]][vertex_to_index[v]] = 1

print("Ma trận kề:")
for row in matrix:
    print(*(row))

print("Danh sách kề:")
for vertex, neighbors in graph.items():
    print(vertex, "->", neighbors)

print("Bậc của từng đỉnh:")
for vertex, neighbors in graph.items():
    print(f"Đỉnh {vertex}: {len(neighbors)}")

print("Nhận xét: Khi đồ thị có ít cạnh, danh sách kề biểu diễn gọn hơn vì chỉ lưu các cạnh thực tế tồn tại, tiết kiệm bộ nhớ so với ma trận kề.")
```
