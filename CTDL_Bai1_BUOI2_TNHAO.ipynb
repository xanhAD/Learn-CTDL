```python
class Stack:
  def __init__(self):
    self._data = []

  def push(self, value):
    self._data.append(value)
  def pop(self):
    if self.is_empty():
      raise IndexError("Stack rỗng")
    return self._data.pop()
  def peek(self):
    if self.is_empty():
      raise IndexError("Stack rỗng")
    return self._data[-1]
  def is_empty(self):
    return len(self._data) == 0

def kiemtra_dau_ngoac(text):
  st = Stack()

  cap_dau = {
      ')': '(',
      ']': '[',
      '}': '{',
  }

  open = "([{"
  close = ")]}"

  for i, txt in enumerate(text):
    if txt in open:
      st.push((txt, i))
    elif txt in close:
      if st.is_empty():
        return False, f"Lỗi vị trí {i}: có '{txt}' nhưng không có dấu mở tương xứng"
      open_txt, open_vitri = st.pop()
      if open_txt != cap_dau[txt]:
        return False, f"Lỗi vị trí {i}: '{txt}' không khớp với '{open_txt}' ở vị trí {open_vitri}"

  if not st.is_empty():
      open_txt, open_vitri = st.peek()
      return False, (
          f"Dấu mở '{open_txt}' tại vị trí {open_vitri} chưa được đóng."
      )

  return True, "Chuỗi hợp lệ."
while True:
  text = input("(Gõ 0 để dừng) Nhập chuỗi: ")
  if text == '0':
    break
  ok, msg = kiemtra_dau_ngoac(text)
  ketqua = "Hợp lệ" if ok else "Không hợp lệ"
  print(text, ketqua, f"->", msg)
```
