```python
from collections import deque

class TreeNode:
    def __init__(self, gia_tri):
        self.val = gia_tri
        self.trai = None
        self.phai = None

def duyet_theo_muc(goc):
    if not goc:
        return []

    ket_qua = []
    hang_doi = deque([goc])

    while hang_doi:
        so_luong = len(hang_doi)
        muc_hien_tai = []

        for _ in range(so_luong):
            nut = hang_doi.popleft()
            muc_hien_tai.append(nut.val)

            if nut.trai:
                hang_doi.append(nut.trai)
            if nut.phai:
                hang_doi.append(nut.phai)

        ket_qua.append(muc_hien_tai)

    return ket_qua

goc = TreeNode(50)
goc.trai = TreeNode(30)
goc.phai = TreeNode(70)

goc.trai.trai = TreeNode(20)
goc.phai.phai = TreeNode(80)

goc.phai.trai = TreeNode(60)
goc.trai.phai = TreeNode(40)

goc.trai.trai.trai = TreeNode(10)

danh_sach = duyet_theo_muc(goc)

for chi_so, muc in enumerate(danh_sach):
    print(f"Mức {chi_so}: {' '.join(map(str, muc))}")

print("Danh sách:", danh_sach)
```
