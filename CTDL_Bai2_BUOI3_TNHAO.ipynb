```python
class TreeNode:
    def __init__(self, gia_tri):
        self.gia_tri = gia_tri
        self.trai = None
        self.phai = None

def dem_Node(root):
    if root is None:
        return 0
    return 1 + dem_Node(root.trai) + dem_Node(root.phai)

def dem_la(root):
    if root is None:
        return 0
    if root.trai is None and root.phai is None:
        return 1
    return dem_la(root.trai) + dem_la(root.phai)

def tong_Node(root):
    if root is None:
        return 0
    return root.gia_tri + tong_Node(root.trai) + tong_Node(root.phai)

def tim_max(root):
    if root is None:
        return None

    max_trai = tim_max(root.trai)
    max_phai = tim_max(root.phai)

    ket_qua = root.gia_tri

    if max_trai is not None and max_trai > ket_qua:
      ket_qua = max_trai
    if max_phai is not None and max_phai > ket_qua:
      ket_qua = max_phai

    return ket_qua

def chieu_cao(root):
    if root is None:
        return -1
    return 1 + max(chieu_cao(root.trai), chieu_cao(root.phai))

root = TreeNode(50)
root.trai = TreeNode(30)
root.phai = TreeNode(70)

root.trai.trai = TreeNode(20)
root.phai.phai = TreeNode(80)

root.phai.trai = TreeNode(60)
root.trai.phai = TreeNode(40)

root.trai.trai.trai = TreeNode(10)

print("Số node: ", dem_Node(root))
print("Số node lá: ", dem_la(root))
print("Tổng giá trị các node: ", tong_Node(root))
print("Giá trị lớn nhất: ", tim_max(root))
print("Chiều cao cây: ", chieu_cao(root))
```
